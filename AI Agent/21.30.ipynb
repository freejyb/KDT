## 22회차 — CrewAI 기본

- CrewAI 공식 구조는 Agent, Task, Crew를 조합하는 방식입니다.

```python
# 핵심:

# Agent = 누구인가?
# Task  = 무슨 일을 하는가?
# Crew  = 누가 함께 일하는가?

# 입니다.
```

```python
%pip install -U crewai
```

```python
from crewai import Agent, Task, Crew


# ==========================================
# 1. Agent
# ==========================================

financial_analyst = Agent(
    role="Financial Analyst",

    goal=(
        "기업의 재무상태와 투자매력도를 "
        "근거 중심으로 분석한다."
    ),

    backstory=(
        "당신은 기업 재무제표와 투자분석을 "
        "전문으로 하는 금융 애널리스트입니다."
    ),

    llm="openai/gpt-4o-mini",

    verbose=True
)


# ==========================================
# 2. Task
# ==========================================

analysis_task = Task(

    description="""
삼성전자를 교육용 사례로 분석하세요.

실제 최신 재무 데이터가 제공되지 않았으므로
존재하지 않는 숫자를 생성하지 마세요.

다음 내용을 작성하세요.

1. 성장성 분석에서 확인할 항목
2. 수익성 분석에서 확인할 항목
3. 안정성 분석에서 확인할 항목
4. 주요 Risk
5. 추가로 필요한 데이터
""",

    expected_output=(
        "구조화된 금융분석 보고서"
    ),

    agent=financial_analyst
)


# ==========================================
# 3. Crew
# ==========================================

crew = Crew(
    agents=[
        financial_analyst
    ],

    tasks=[
        analysis_task
    ],

    verbose=True
)


# ==========================================
# 4. 실행
# ==========================================

result = crew.kickoff()

print("\n======================")
print("CrewAI 결과")
print("======================")

print(result)
```

## 23회차 — CrewAI Multi-Agent

- 이번에는 실제 투자분석팀을 만듭니다.
- Market Analyst
- Fundamental Analyst
- Research Analyst
- 
        ↓
 - Portfolio Manager

```python
from crewai import (
    Agent,
    Task,
    Crew,
    Process
)


# ==========================================
# Agents
# ==========================================

market_analyst = Agent(

    role="Market Analyst",

    goal=(
        "주가와 시장 위험 관점에서 "
        "기업을 분석한다."
    ),

    backstory=(
        "주식시장과 Quant 분석을 담당하는 "
        "Market Analyst"
    ),

    llm="openai/gpt-4o-mini"
)


fundamental_analyst = Agent(

    role="Fundamental Analyst",

    goal=(
        "기업의 재무건전성과 "
        "사업성과를 분석한다."
    ),

    backstory=(
        "재무제표와 Valuation 전문 "
        "Fundamental Analyst"
    ),

    llm="openai/gpt-4o-mini"
)


research_analyst = Agent(

    role="Research Analyst",

    goal=(
        "기업 분석에 필요한 뉴스·공시·"
        "산업정보를 구조화한다."
    ),

    backstory=(
        "공시와 산업정보 조사 전문 Analyst"
    ),

    llm="openai/gpt-4o-mini"
)


portfolio_manager = Agent(

    role="Portfolio Manager",

    goal=(
        "전문 Analyst의 결과를 종합하여 "
        "균형 잡힌 Investment Thesis를 만든다."
    ),

    backstory=(
        "투자위원회를 총괄하는 "
        "Portfolio Manager"
    ),

    llm="openai/gpt-4o-mini"
)


# ==========================================
# Tasks
# ==========================================

market_task = Task(

    description="""
삼성전자에 대해 Market Analyst 관점에서
분석할 항목을 작성하세요.

- 주가
- 수익률
- 변동성
- MDD
- Benchmark

실제 수치는 만들지 마세요.
""",

    expected_output="Market Analysis",

    agent=market_analyst
)


fundamental_task = Task(

    description="""
삼성전자에 대해 Fundamental 관점에서
분석할 항목을 정리하세요.

- 매출
- 이익
- ROE
- Cash Flow
- 부채
- Valuation

제공되지 않은 수치는 만들지 마세요.
""",

    expected_output="Fundamental Analysis",

    agent=fundamental_analyst
)


research_task = Task(

    description="""
삼성전자의 Research Agent라면
어떤 뉴스·공시·산업정보를 조사해야 하는지
정리하세요.

실제 최신 뉴스는 생성하지 마세요.
""",

    expected_output="Research Analysis",

    agent=research_analyst
)


manager_task = Task(

    description="""
세 Analyst의 결과를 종합하여

1. Executive Summary
2. 투자 포인트
3. Risk
4. 추가 필요 데이터
5. Investment Thesis

를 작성하세요.
""",

    expected_output="Equity Research Report",

    agent=portfolio_manager,

    context=[
        market_task,
        fundamental_task,
        research_task
    ]
)


# ==========================================
# Crew
# ==========================================

crew = Crew(

    agents=[
        market_analyst,
        fundamental_analyst,
        research_analyst,
        portfolio_manager
    ],

    tasks=[
        market_task,
        fundamental_task,
        research_task,
        manager_task
    ],

    process=Process.sequential,

    verbose=True
)


result = crew.kickoff()

print(result)
```

## 24회차 — AutoGen Investment Committee

- AutoGen 최신 AgentChat에서는 AssistantAgent와 RoundRobinGroupChat을 이용할 수 있습니다.

- 이 회차의 포인트는 Agent가 서로의 출력을 읽고 반론한다는 것입니다.

```python
# %pip install -U autogen-agentchat "autogen-ext[openai]"
```

```python
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.teams import RoundRobinGroupChat
from autogen_agentchat.conditions import (
    TextMentionTermination,
    MaxMessageTermination
)
from autogen_agentchat.ui import Console

from autogen_ext.models.openai import (
    OpenAIChatCompletionClient
)


# ==========================================
# Model
# ==========================================

model_client = OpenAIChatCompletionClient(
    model="gpt-4o-mini"
)


# ==========================================
# Bull Analyst
# ==========================================

bull = AssistantAgent(

    name="bull_analyst",

    model_client=model_client,

    system_message="""
당신은 Bull Analyst입니다.

기업의 투자매력과 긍정적인 요소를 찾되
제공되지 않은 숫자나 최신 사실을 만들지 마세요.
"""
)


# ==========================================
# Bear Analyst
# ==========================================

bear = AssistantAgent(

    name="bear_analyst",

    model_client=model_client,

    system_message="""
당신은 Bear Analyst입니다.

투자 아이디어의 약점과
하방 위험을 중심으로 반론하세요.
"""
)


# ==========================================
# Risk Analyst
# ==========================================

risk = AssistantAgent(

    name="risk_analyst",

    model_client=model_client,

    system_message="""
당신은 Risk Analyst입니다.

Bull과 Bear 논리에서 놓친
시장·산업·재무 Risk를 검토하세요.
"""
)


# ==========================================
# Portfolio Manager
# ==========================================

manager = AssistantAgent(

    name="portfolio_manager",

    model_client=model_client,

    system_message="""
당신은 Portfolio Manager입니다.

앞선 Analyst의 의견을 종합하세요.

마지막 문장은 반드시

FINAL

로 끝내세요.
"""
)


# ==========================================
# 종료 조건
# ==========================================

termination = (
    TextMentionTermination("FINAL")
    |
    MaxMessageTermination(max_messages=8)
)


# ==========================================
# Team
# ==========================================

team = RoundRobinGroupChat(

    [
        bull,
        bear,
        risk,
        manager
    ],

    termination_condition=termination
)


# ==========================================
# 실행
# ==========================================

await Console(
    team.run_stream(

        task="""
삼성전자를 교육용 사례로
투자위원회 토론을 진행하세요.

실제 최신 데이터가 없으므로
사실이나 숫자를 임의 생성하지 마세요.
"""
    )
)


await model_client.close()
```

## 25회차 — LlamaIndex + RAG

- LlamaIndex에서는 문서를 VectorStoreIndex에 넣고 QueryEngine으로 검색할 수 있습니다.

```python
%pip install -U \
    llama-index \
    llama-index-llms-openai \
    llama-index-embeddings-openai
```

- 금융문서 샘플 RAG

- 질문
↓
- 검색
 ↓
- Evidence
 ↓
- LLM
 ↓
- Answer

```python
from llama_index.core import (
    Document,
    VectorStoreIndex,
    Settings
)

from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import (
    OpenAIEmbedding
)


# ==========================================
# 1. LLM / Embedding
# ==========================================

Settings.llm = OpenAI(
    model="gpt-4o-mini",
    temperature=0
)

Settings.embed_model = OpenAIEmbedding(
    model="text-embedding-3-small"
)


# ==========================================
# 2. 금융 문서
# ==========================================

documents = [

    Document(
        text="""
[사업보고서]
A기업은 반도체 사업의 매출 비중이 높다.
산업 경기 변화에 따라 실적 변동성이 커질 수 있다.
"""
    ),

    Document(
        text="""
[IR 자료]
A기업은 AI 데이터센터 관련
고성능 제품 투자를 확대하고 있다.
신규 시장 확대는 성장 기회가 될 수 있다.
"""
    ),

    Document(
        text="""
[Risk 자료]
환율 변동, 글로벌 경기 둔화,
설비투자 증가가 주요 위험요인이다.
"""
    )
]


# ==========================================
# 3. Index
# ==========================================

index = VectorStoreIndex.from_documents(
    documents
)


# ==========================================
# 4. Query Engine
# ==========================================

query_engine = index.as_query_engine(
    similarity_top_k=2
)


# ==========================================
# 5. 질문
# ==========================================

response = query_engine.query(
    """
A기업의 투자 관점에서
가장 중요한 성장요인과 Risk를
근거를 바탕으로 정리해줘.
"""
)


print("\n=== RAG 답변 ===")

print(response)


# ==========================================
# 6. Retrieval 근거 확인
# ==========================================

print("\n=== 검색된 근거 ===")

for i, source in enumerate(
    response.source_nodes,
    start=1
):

    print(
        f"\nSource {i}"
    )

    print(
        "Score:",
        source.score
    )

    print(
        source.node.text
    )
```

## 26회차 — Low-code Financial Workflow

- Dify, Langflow, n8n 자체는 주로 UI에서 Node를 연결하는 과정

```python
from langchain_openai import ChatOpenAI


llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)


# ==========================================
# Node 1
# Market API 역할
# ==========================================

def market_node(company):

    return {
        "company": company,

        "market": (
            "교육용 샘플: "
            "주가·수익률·변동성 분석 필요"
        )
    }


# ==========================================
# Node 2
# DART 역할
# ==========================================

def financial_node(state):

    state["financial"] = (
        "매출·영업이익·ROE·"
        "현금흐름 분석 필요"
    )

    return state


# ==========================================
# Node 3
# News 역할
# ==========================================

def news_node(state):

    state["news"] = (
        "산업 뉴스·공시·Catalyst "
        "확인 필요"
    )

    return state


# ==========================================
# Node 4
# LLM Report
# ==========================================

def report_node(state):

    prompt = f"""
당신은 금융 애널리스트입니다.

기업:
{state['company']}

Market:
{state['market']}

Financial:
{state['financial']}

News:
{state['news']}

위 정보를 이용해
교육용 투자분석 보고서를 작성하세요.
"""

    response = llm.invoke(
        prompt
    )

    state["report"] = (
        response.content
    )

    return state


# ==========================================
# Workflow
# ==========================================

state = market_node(
    "삼성전자"
)

state = financial_node(
    state
)

state = news_node(
    state
)

state = report_node(
    state
)


print(
    state["report"]
)
```

## 27회차 — Quant + LLM Hybrid Agent

- 숫자는 Python, 해석은 LLM에게 맡깁니다. --> Python은 계산하고, LLM은 해석한다.

```python
%pip install -U pandas numpy langchain-openai
```

```python
import numpy as np
import pandas as pd

from langchain_openai import (
    ChatOpenAI
)


# ==========================================
# 1. 샘플 주가 생성
# ==========================================

np.random.seed(42)

daily_returns = np.random.normal(
    loc=0.0005,
    scale=0.015,
    size=252
)


prices = (
    100
    *
    np.cumprod(
        1 + daily_returns
    )
)


df = pd.DataFrame({
    "price": prices
})


# ==========================================
# 2. 수익률
# ==========================================

df["return"] = (
    df["price"].pct_change()
)


total_return = (
    df["price"].iloc[-1]
    /
    df["price"].iloc[0]
    - 1
)


# ==========================================
# 3. 변동성
# ==========================================

volatility = (
    df["return"].std()
    *
    np.sqrt(252)
)


# ==========================================
# 4. MDD
# ==========================================

running_max = (
    df["price"].cummax()
)

drawdown = (
    df["price"]
    /
    running_max
    - 1
)

mdd = drawdown.min()


print(
    "Total Return:",
    round(
        total_return * 100,
        2
    ),
    "%"
)

print(
    "Volatility:",
    round(
        volatility * 100,
        2
    ),
    "%"
)

print(
    "MDD:",
    round(
        mdd * 100,
        2
    ),
    "%"
)


# ==========================================
# 5. LLM 해석
# ==========================================

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)


response = llm.invoke(
    f"""
다음 금융지표를
투자자 관점에서 해석하세요.

총수익률:
{total_return:.2%}

연환산 변동성:
{volatility:.2%}

MDD:
{mdd:.2%}

규칙:
- 계산값은 변경하지 마세요.
- 수익성과 위험을 함께 설명하세요.
- 매수/매도를 단정하지 마세요.
"""
)


print(
    "\n=== LLM Analysis ==="
)

print(
    response.content
)
```

```python
# 현재 MCP Python SDK v2는 Python 3.10 이상을 요구합니다. 
#     따라서 28회차는 Python 3.10+ 환경을 별도로 만드는 것이 좋습니다.
```

```python
from mcp.server import MCPServer


mcp = MCPServer(
    "Financial Analysis Server"
)


# ==========================================
# Tool 1
# ==========================================

@mcp.tool()
def get_stock_summary(
    company: str
) -> dict:
    """
    교육용 주가 요약을 반환한다.
    """

    return {
        "company": company,
        "return": "8%",
        "volatility": "20%",
        "note": "교육용 샘플"
    }


# ==========================================
# Tool 2
# ==========================================

@mcp.tool()
def get_financial_ratio(
    company: str
) -> dict:
    """
    교육용 재무비율을 반환한다.
    """

    return {
        "company": company,
        "ROE": "12%",
        "debt_ratio": "35%",
        "note": "교육용 샘플"
    }


# ==========================================
# Resource
# ==========================================

@mcp.resource(
    "finance://policy"
)
def finance_policy() -> str:

    return """
금융분석 정책

1. 없는 숫자를 생성하지 않는다.
2. 데이터 출처를 남긴다.
3. 분석 기준일을 표시한다.
4. 투자 판단을 과도하게 단정하지 않는다.
"""


# ==========================================
# 실행
# ==========================================

if __name__ == "__main__":
    mcp.run()
```

```python
import numpy as np
import pandas as pd

from crewai import (
    Agent,
    Task,
    Crew,
    Process
)

from llama_index.core import (
    Document,
    VectorStoreIndex,
    Settings
)

from llama_index.llms.openai import (
    OpenAI
)

from llama_index.embeddings.openai import (
    OpenAIEmbedding
)


# ==========================================
# 1. Quant 분석
# ==========================================

def quant_analysis():

    np.random.seed(42)

    returns = np.random.normal(
        0.0005,
        0.015,
        252
    )

    prices = (
        100
        *
        np.cumprod(
            1 + returns
        )
    )

    series = pd.Series(
        prices
    )

    total_return = (
        series.iloc[-1]
        /
        series.iloc[0]
        - 1
    )

    volatility = (
        pd.Series(
            returns
        ).std()
        *
        np.sqrt(252)
    )

    drawdown = (
        series
        /
        series.cummax()
        - 1
    )

    mdd = drawdown.min()

    return f"""
교육용 Quant 결과

총수익률:
{total_return:.2%}

연환산 변동성:
{volatility:.2%}

MDD:
{mdd:.2%}
"""


quant_result = (
    quant_analysis()
)


print(
    quant_result
)


# ==========================================
# 2. RAG
# ==========================================

Settings.llm = OpenAI(
    model="gpt-4o-mini",
    temperature=0
)

Settings.embed_model = (
    OpenAIEmbedding(
        model="text-embedding-3-small"
    )
)


documents = [

    Document(
        text="""
사업보고서:
반도체 산업은 경기 민감도가 높아
실적 변동성이 존재할 수 있다.
"""
    ),

    Document(
        text="""
IR 자료:
AI 데이터센터와 고성능 반도체 시장은
중장기 성장 기회가 될 수 있다.
"""
    ),

    Document(
        text="""
Risk 자료:
글로벌 경기 둔화,
환율 변화,
대규모 설비투자가
주요 위험요인이다.
"""
    )
]


index = (
    VectorStoreIndex
    .from_documents(
        documents
    )
)


query_engine = (
    index.as_query_engine(
        similarity_top_k=3
    )
)


rag_response = query_engine.query(
    """
기업의 주요 성장기회와
Risk를 정리하세요.
"""
)


rag_result = str(
    rag_response
)


print(
    "\n=== RAG ==="
)

print(
    rag_result
)


# ==========================================
# 3. CrewAI Agents
# ==========================================

market_agent = Agent(

    role="Market Analyst",

    goal=(
        "Quant 결과를 정확하게 "
        "투자 관점으로 해석한다."
    ),

    backstory=(
        "시장위험과 Quant 분석 전문가"
    ),

    llm="openai/gpt-4o-mini"
)


research_agent = Agent(

    role="Research Analyst",

    goal=(
        "RAG Evidence를 바탕으로 "
        "성장요인과 위험을 분석한다."
    ),

    backstory=(
        "공시·IR·산업자료 조사 전문가"
    ),

    llm="openai/gpt-4o-mini"
)


critic_agent = Agent(

    role="Critic",

    goal=(
        "근거 없는 숫자와 "
        "과도한 투자 판단을 찾는다."
    ),

    backstory=(
        "금융보고서 검증 전문가"
    ),

    llm="openai/gpt-4o-mini"
)


manager = Agent(

    role="Portfolio Manager",

    goal=(
        "모든 분석을 통합하여 "
        "균형 잡힌 투자보고서를 만든다."
    ),

    backstory=(
        "투자위원회를 총괄하는 "
        "Portfolio Manager"
    ),

    llm="openai/gpt-4o-mini"
)


# ==========================================
# 4. Tasks
# ==========================================

market_task = Task(

    description=f"""
다음 Quant 결과를 해석하세요.

{quant_result}

규칙:
숫자를 변경하거나
새로운 숫자를 만들지 마세요.
""",

    expected_output=(
        "Market Analysis"
    ),

    agent=market_agent
)


research_task = Task(

    description=f"""
다음 RAG 결과를 이용해
Growth Catalyst와 Risk를 분석하세요.

{rag_result}

근거가 없는 사실은
추가하지 마세요.
""",

    expected_output=(
        "Evidence 기반 Research"
    ),

    agent=research_agent
)


critic_task = Task(

    description="""
Market과 Research 결과를 검토하세요.

다음을 확인하세요.

1. 근거 없는 숫자
2. 근거 없는 사실
3. Risk 누락
4. 과도한 투자 단정
""",

    expected_output=(
        "Critic 검증 결과"
    ),

    agent=critic_agent,

    context=[
        market_task,
        research_task
    ]
)


manager_task = Task(

    description="""
Market Analysis,
Research,
Critic 결과를 통합하여

다음 목차로 최종 보고서를 작성하세요.

1. Executive Summary
2. Market Analysis
3. Fundamental / Research View
4. Growth Catalyst
5. Risk
6. Investment Thesis
7. 추가 필요 데이터

교육용 데이터라는 점을
명확히 표시하세요.
""",

    expected_output=(
        "Final Equity Research Report"
    ),

    agent=manager,

    context=[
        market_task,
        research_task,
        critic_task
    ]
)


# ==========================================
# 5. Crew
# ==========================================

crew = Crew(

    agents=[
        market_agent,
        research_agent,
        critic_agent,
        manager
    ],

    tasks=[
        market_task,
        research_task,
        critic_task,
        manager_task
    ],

    process=Process.sequential,

    verbose=True
)


# ==========================================
# 6. 실행
# ==========================================

final_result = (
    crew.kickoff()
)


print(
    "\n================================"
)

print(
    "FINAL INVESTMENT RESEARCH REPORT"
)

print(
    "================================"
)

print(
    final_result
)
```
