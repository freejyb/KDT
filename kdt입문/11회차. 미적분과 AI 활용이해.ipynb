## 경사하강법으로 최적 가중치 찾기

```python
# 실습 목표 ;  다음 손실함수를 최소화하는 가중치 w를 찾습니다.
# 손실함수: L(w) = (w - 3)²
# 이 함수는 w = 3일 때 손실이 0으로 가장 작습니다.
# 미분값은 다음과 같습니다.
# 기울기: dL/dw = 2(w - 3)
```

### 1단계: 손실함수와 기울기 정의

```python
# 손실함수
def loss_function(w):
    return (w - 3) ** 2

# 손실함수의 미분값
def gradient(w):
    return 2 * (w - 3)
```

```python
# loss_function(w)는 현재 가중치의 오차를 계산합니다.
# gradient(w)는 가중치를 어느 방향으로 바꿔야 손실이 감소하는지 알려줍니다.
```

### 2단계: 경사하강법 실행

```python
# 초기 가중치
w = 0

# 학습률
learning_rate = 0.1

# 반복 횟수
epochs = 10

for epoch in range(epochs):
    loss = loss_function(w)
    grad = gradient(w)

    print(
        f"{epoch + 1}회: "
        f"w={w:.4f}, "
        f"기울기={grad:.4f}, "
        f"손실={loss:.4f}"
    )

    # 가중치 업데이트
    w = w - learning_rate * grad

print(f"\n최종 가중치: {w:.4f}")
```

```python
# 반복 횟수를 늘리면 w는 최솟값인 3에 점점 가까워집니다.
# # 반복 횟수: epochs = 50
```

```python
# 초기 가중치
w = 0

# 학습률
learning_rate = 0.1

# 반복 횟수
epochs = 50

for epoch in range(epochs):
    loss = loss_function(w)
    grad = gradient(w)

    print(
        f"{epoch + 1}회: "
        f"w={w:.4f}, "
        f"기울기={grad:.4f}, "
        f"손실={loss:.4f}"
    )

    # 가중치 업데이트
    w = w - learning_rate * grad

print(f"\n최종 가중치: {w:.4f}")
```

### 3단계: 손실이 감소하는 과정 시각화

```python
import matplotlib.pyplot as plt

w = 0
learning_rate = 0.1
epochs = 20

weight_history = []
loss_history = []

for epoch in range(epochs):
    weight_history.append(w)
    loss_history.append(loss_function(w))

    grad = gradient(w)
    w = w - learning_rate * grad

plt.figure(figsize=(8, 4))
plt.plot(range(1, epochs + 1), loss_history, marker="o")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Gradient Descent: Loss Reduction")
plt.grid()
plt.show()
```

```python
# 그래프 해석
# 학습 초기에는 손실이 크게 감소합니다.
# 최솟값에 가까워질수록 손실 감소 폭이 작아집니다.
# 경사하강법이 손실을 점차 줄여가는 모습을 확인할 수 있습니다.
```

### 4단계: 학습률 비교 실습

```python
learning_rates = [0.01, 0.1, 0.9]

for learning_rate in learning_rates:
    w = 0

    print(f"\n학습률: {learning_rate}")

    for epoch in range(5):
        grad = gradient(w)
        w = w - learning_rate * grad

        print(
            f"{epoch + 1}회: "
            f"w={w:.4f}, "
            f"손실={loss_function(w):.4f}"
        )
```

```python
# 학습률	결과
# 0.01	안정적이지만 매우 느리게 이동
# 0.10	비교적 빠르고 안정적으로 최솟값에 접근
# 0.90	최솟값 주변을 크게 오가며 불안정할 수 있음
```
