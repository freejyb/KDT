## 내적·포트폴리오·행렬곱·유사도

### 1. 벡터 덧셈과 뺄셈

```python
import numpy as np
import pandas as pd
```

```python
january_returns = np.array([0.02, -0.01, 0.03])
february_returns = np.array([0.01, 0.02, -0.02])

print("벡터 덧셈:", january_returns + february_returns)
print("벡터 뺄셈:", february_returns - january_returns)
```

### 2. 스칼라배

```python
investment = np.array([100, 200, 300])

double_investment = 2 * investment

print("기존 투자금액:", investment)
print("2배 투자금액:", double_investment)
```

### 3. 벡터 내적 계산

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_product = np.dot(a, b)

print("내적:", dot_product)
```

```python
# 계산 과정:
# 1×4 + 2×5 + 3×6
# = 4 + 10 + 18
# = 32
```

```python
# @ 연산자를 사용할 수도 있습니다.

print("내적:", a @ b)
```

### 4. 포트폴리오 수익률 계산

```python
# 삼성전자, 현대차, 네이버 투자비중
weights = np.array([0.5, 0.3, 0.2])

# 각 자산 수익률
asset_returns = np.array([0.04, 0.02, -0.01])

portfolio_return = weights @ asset_returns

print(f"포트폴리오 수익률: {portfolio_return * 100:.2f}%")
```

```python
# 계산 과정:
# 0.5×0.04 + 0.3×0.02 + 0.2×(-0.01)
# = 0.024
# = 2.4%
```

```python
# f" ": f-string
# 문자열 앞의 f는 문자열 안에 변수나 계산 결과를 넣을 수 있게 합니다.
# :.2f
# 숫자를 소수점 아래 둘째 자리까지 표시하라는 형식 지정입니다.
# .2 → 소수점 아래 2자리
# f  → 실수 형식
```

### 5. 수익률 행렬과 투자비중의 행렬곱

```python
# 행: 날짜
# 열: 삼성전자, 현대차, 네이버

return_matrix = np.array([
    [ 0.01,  0.02, -0.01],
    [ 0.03, -0.01,  0.02],
    [-0.02,  0.01,  0.00]
])

daily_portfolio_returns = return_matrix @ weights

print("날짜별 포트폴리오 수익률:")
print(daily_portfolio_returns)
```

```python
# 백분률로 출력
for day, return_rate in enumerate(daily_portfolio_returns, start=1):
    print(f"{day}일차 포트폴리오 수익률: {return_rate * 100:.2f}%")
```

### 6. 코사인 유사도 계산

```python
samsung = np.array([0.01, -0.02, 0.005])
sk_hynix = np.array([0.012, -0.018, 0.007])
kepco = np.array([-0.003, 0.001, -0.002])

def cosine_similarity(vector_a, vector_b):
    dot_product = np.dot(vector_a, vector_b)
    size_a = np.linalg.norm(vector_a)
    size_b = np.linalg.norm(vector_b)

    return dot_product / (size_a * size_b)

similarity_1 = cosine_similarity(samsung, sk_hynix)
similarity_2 = cosine_similarity(samsung, kepco)

print(f"삼성전자-SK하이닉스 유사도: {similarity_1:.4f}")
print(f"삼성전자-한국전력 유사도: {similarity_2:.4f}")
```

```python
# 해석
# 1에 가까움  → 수익률 움직임이 매우 유사
# 0에 가까움  → 방향의 관련성이 낮음
# -1에 가까움 → 반대 방향으로 움직이는 경향
```

### 7. AI 입력값과 가중치의 내적

```python
# 입력값: 부채비율, 유동비율, 영업이익률
x = np.array([180, 120, 8])

# 각 변수의 가중치
w = np.array([0.04, -0.02, -0.10])

# 절편
b = -3

score = x @ w + b

print("AI 모형의 부실점수:", score)
```

```python
# 계산 구조:
# 부실점수
# = 180×0.04
# + 120×(-0.02)
# + 8×(-0.10)
# - 3
```

### 8. 여러 기업의 부실점수를 한 번에 계산

```python
X = np.array([
    [180, 120,  8],
    [300,  70, -2],
    [100, 160, 12]
])

w = np.array([0.04, -0.02, -0.10])
b = -3

scores = X @ w + b

print("기업별 부실점수:")
print(scores)
```

```python
#기업명과 함께 출력
company_names = ["A기업", "B기업", "C기업"]

for company, score in zip(company_names, scores):
    print(f"{company}: {score:.2f}")
```
