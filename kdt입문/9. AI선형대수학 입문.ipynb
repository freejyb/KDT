## 벡터·행렬과 금융데이터 표현

### 1. 라이브러리 불러오기

```python
import numpy as np
import pandas as pd
```

### 2. 스칼라와 벡터 만들기

```python
# 스칼라: 하나의 숫자
stock_price = 70000
interest_rate = 0.035

print("삼성전자 주가:", stock_price)
print("기준금리:", interest_rate)
```

```python
# 벡터: 여러 숫자를 순서대로 저장
prices = np.array([70000, 71000, 69500, 72000])

print("주가 벡터:", prices)
print("벡터의 차원:", prices.ndim)
print("벡터의 원소 개수:", prices.shape[0])
```

```python
# 해석
# prices.ndim
# → 배열의 축 개수

# prices.shape
# → 각 축에 포함된 원소 개수

# prices는 원소가 4개인 1차원 배열입니다.
# 금융데이터 관점에서는 4개 날짜의 주가를 가진 벡터입니다.
```

### 3. 기업 재무정보를 벡터로 표현

```python
# 부채비율, 유동비율, 영업이익률
company_a = np.array([180, 120, 8])

print("A기업 재무벡터:", company_a)
print("특성의 개수:", len(company_a))
```

```python
# 결과:

# A기업 재무벡터: [180 120   8]
# 특성의 개수: 3

# A기업은 3개의 재무변수로 표현되므로 3차원 특성벡터입니다.
```

### 4. 벡터의 크기 계산

```python
v = np.array([3, 4])

vector_size = np.linalg.norm(v)

print("벡터:", v)
print("벡터의 크기:", vector_size)
```

```python
# 계산 원리:

# 벡터의 크기
# = √(3² + 4²)
# = 5
```

### 5. 금융데이터 행렬 만들기

```python
# 행: 기업
# 열: 부채비율, 유동비율, 영업이익률

financial_matrix = np.array([
    [180, 120,  8],
    [300,  70, -2],
    [100, 160, 12]
])

print("금융데이터 행렬:")
print(financial_matrix)

print("행렬 크기:", financial_matrix.shape)
print("행의 개수:", financial_matrix.shape[0])
print("열의 개수:", financial_matrix.shape[1])
```

```python
# 결과의 의미:

# 행렬 크기: (3, 3)

# 3개 기업 × 3개 재무변수
```

### 6. 전치행렬 확인

```python
transpose_matrix = financial_matrix.T

print("원래 행렬:")
print(financial_matrix)

print("\n전치행렬:")
print(transpose_matrix)
```

```python
# 원래 행렬에서는 행이 기업이고 열이 변수입니다.
# 전치하면 다음과 같이 바뀝니다.
# 행: 재무변수
# 열: 기업
```

### 7. Pandas DataFrame으로 표현

```python
df = pd.DataFrame(
    financial_matrix,
    index=["A기업", "B기업", "C기업"],
    columns=["부채비율", "유동비율", "영업이익률"]
)

df
```

```python
# DataFrame 정보 확인
print("DataFrame 크기:", df.shape)
print("행 이름:", df.index.tolist())
print("열 이름:", df.columns.tolist())
```

### 8. DataFrame을 행렬로 변환

```python
X = df.to_numpy()

print(X)
print(type(X))
```

```python
# 핵심 설명
# Pandas DataFrame
# → 변수명과 기업명이 포함된 분석용 표
# NumPy 배열
# → AI가 수학 계산에 활용하는 숫자 행렬
```
