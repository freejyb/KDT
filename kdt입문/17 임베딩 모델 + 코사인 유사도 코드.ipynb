### question = "최고 금리는 얼마인가요?"

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

chunks = [
    "스마트 정기예금의 가입 기간은 12개월입니다.",
    "기본금리는 연 2.8%이고 우대금리는 최대 연 0.5%포인트입니다.",
    "우대 조건은 급여이체, 카드 사용, 모바일 가입입니다."
]

question = "최고 금리는 얼마인가요?"

model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

chunk_vectors = model.encode(chunks)
question_vector = model.encode([question])

scores = cosine_similarity(question_vector, chunk_vectors)[0]

for i, score in enumerate(scores):
    print(f"Chunk {i+1}: {score:.3f}")

best = scores.argmax()
print("\n검색된 문서:", chunks[best])
```
