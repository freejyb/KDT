# 라이브러리 (Library)

## 라이브러리 란?

파이썬 라이브러리(Python Library)는 특정 기능을 쉽게 구현할 수 있도록 미리 작성된 파이썬 코드들의 집합

쉽게 말해, 프로그래밍에서 자주 사용하는 기능을 묶어놓은 도구 상자라고 생각하면 됩니다.

라이브러리의 역할
코드 재사용: 매번 직접 코드를 작성하지 않고, 필요한 기능을 불러와서 바로 사용 가능.

개발 효율성 향상: 데이터 분석, 웹 개발, 인공지능 등 다양한 분야의 복잡한 기능을 손쉽게 구현

표준화된 기능 제공: 여러 개발자가 공통적으로 사용하는 검증된 기능을 제공하여 안정성과 신뢰성을 높임

라이브러리의 종류
파이썬 라이브러리는 크게 두 가지로 나눌 수 있습니다.

(1) 표준 라이브러리 (Standard Library)

파이썬을 설치하면 자동으로 제공되는 기본 라이브러리입니다.

예를 들어:

math → 수학 계산 함수 제공 (sqrt, sin, log 등)

datetime → 날짜와 시간 처리

os → 운영체제와 상호작용 (파일 경로, 디렉터리 관리 등)

json → JSON 데이터 읽기/쓰기

(2) 외부 라이브러리 (Third-party Library)

파이썬 기본 설치에는 포함되지 않지만, pip 명령어를 사용해 설치 후 활용하는 라이브러리입니다.

pip(Package Installer for Python)

예를 들어:

데이터 분석: pandas, numpy

시각화: matplotlib, seaborn

머신러닝/딥러닝: scikit-learn, tensorflow, pytorch

웹 개발: flask, django

웹 크롤링: requests, beautifulsoup4

```python
# 제어문(조건문, 반복문) : 조건문이나 반복문처럼 프로그램 명령어가 기술된 순서대로 순차 수행하는 경우
# 조건문 : if, 반복문 : for, while
```

```python
# 조건문(if문) : 어떤 조건을 보고 코드를 수행하는 것 ==> if ,  if~else , if~elif ~ else
# 반복문(for,  while) : 어떤 조건하에서 반복적으로 코드를 수행하는 것 
```

```python
# if 조건문
# 파이썬에서 조건문은 어떤 조건이 참(True)일 때 특정 코드를 실행하도록 하는 구문입니다.
# 가장 기본적인 형태는 if, elif, else 문입니다.
# if 조건식: # 조건식이 참일 때 실행할 코드
# elif 다른_조건식: # 다른_조건식이 참일 때 실행할 코드
# else: # 위의 조건들이 모두 거짓일 때 실행할 코드
# if : 조건문을 표현
# if 문의 끝에는 콜론(:)을 입력
# if 문의 조건이 True일 때 실행되는 문장은 들여쓰기
```

```python
return_rate = 8

if return_rate > 0:
    print("수익입니다.")
elif return_rate == 0:
    print("손익이 없습니다.")
else:
    print("손실입니다.")
```

```python
# 사례 1. 수익·손실 판단


buy_price = 50000
sell_price = 55000

return_rate = (sell_price - buy_price) / buy_price * 100

if return_rate > 0:
    print("수익")
elif return_rate == 0:
    print("손익 없음")
else:
    print("손실")
```

```python
# 사례 2. 목표수익률 판단
target_rate = 10

if return_rate >= target_rate:
    print("목표수익률 달성")
else:
    print("목표수익률 미달")
```
