```python
import numpy as np
from scipy.optimize import minimize

# 목적함수
def objective(x):
    return (x[0] - 3)**2 + (x[1] - 2)**2

# 제약조건: x + y <= 4
constraints = [{
    'type': 'ineq',
    'fun': lambda x: 4 - x[0] - x[1]
}]

# x, y >= 0
bounds = [(0, None), (0, None)]

# 초기값
x0 = [0, 0]

# SLSQP 최적화
result = minimize(
    objective,
    x0,
    method='SLSQP',
    bounds=bounds,
    constraints=constraints
)

print("최적해:", result.x)
print("최소 목적함수 값:", result.fun)
```
